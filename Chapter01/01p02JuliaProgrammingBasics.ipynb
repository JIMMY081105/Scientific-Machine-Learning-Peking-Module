# Types

As with many programming languages, variables in Julia can have different types.  By default, integers have type *Int64*, and float is double precision *Float64*.  There is also a string type to represent a string. You can set values to  variables in Julia by just using the = sign

```julia
x=1
y=2.0
z="Here is a string"
```

You can find out the type of the variables by using the `typeof()` function in Julia

```julia
typeof(x)
```

Note that when we set `x` to the value of 1, Julia by default set x to be an integer of type *Int64*

<u>Exercise Ex01:</u>  Use `typeof()` to find out the type of variables `y` and `z`

You can force a variable to be single precision floating point with the code below

```julia
p=1.0f0
q=Float32(1.0)
typeof(p),typeof(q)
```

You can group variables together by using a `tuple` in Julia

```julia
t=(x,y,z)
```

Here the tuple `t` has three elements an integer, float and a string.  You can also use the `typeof()` function to find out the type of variables in t

```julia
typeof(t)
```

Note that tuples are immutable, this means that their values are set and cannot be changed after the creation of the tuple.  You can access each element of the tuple by using the square bracket [] with a number indicating the position of the element in the tuple.

```julia
t[3]
```

Another way of grouping different variables together is by using a `struct`.  

```julia
struct House
    Address::String
    Size::Float32
    Construction::String
end

MyHouse=House("Australia",100.0,"Brick")
```

You can access each element of the struct by using the . notation

```julia
MyHouse.Address
```

By default structs are immutable, that means you cannot change the variables in struct.  The command

```julia
MyHouse.Size=200
```

If you want to be able to change the elements inside struct, you will need to use a mutable struct 

```julia
# Define a mutable struct called Counter
mutable struct Age
    Years::Int
end
MyAge=Age(70)
```

```julia
MyAge.Years=60
```

```julia
MyAge.Years
```

# Functions

Functions are used to group codes that is repeatedly used.  Simple functions in Julia can be defined by a one line syntax.

```julia
f(x)=x^3
```

```julia
typeof(f(1)),typeof(f(2.0))
```

Note that the type of the return value of $f$ depends of the input argument type, $x$.  If $x$ is an integer, then $f$ returns an integer.  If $x$ is an float, then $f$ returns an float.

You can make f act on every element of a by using the . operator

```julia
f.(a)
```

<u>Exercise Ex06:</u>  What does `f(a)` give you?  Why?

More complex functions can be defined by using the `function` keyword.  The last expression in the function is the return value by default

```julia
function my_first_function(x,y)
    mean=0.5*(x+y)
end

my_first_function(1.0,2.0)
```

One can define a function of two variables and then map it to a function of one variable.  This basically ignores the other parameter.  This is common in the optimization packages in Julia which will be used in future notebooks.


In the example below. `dummy_func()` is declared as function of two variables `x` and `p`.  But it returns $f(x)=x^3$, i.e. only `x`  is used and `p` is ignored.

```julia
dummy_func=(x,p)->f(x)
dummy_func(3,10) 
```

# Scope of variables

Usually variables inside functions are local

```julia
x=2
function my_print(x)
    println(x)
end

my_print(3)
```

In the code above, the variable *x* is local inside the `my_print()` function

However, if Julia does not see a variable inside a function, then it will assume that the variable is global and try to find the value of the variable outside the function.  

```julia
x=2
function my_print(b)
    println(x)
end

my_print(10)
```

In the code above, *x* is not defined inside `my_print()` so when `println()` will find try to find the variable x outside the function which has a value of 2.

# Loops

We usually use loops to do repeated calculations.  For example the `for` loop below takes away 1 from the variable *x* 5 times

```julia
x=7.0
for i=1:5
    x-=1.0
end
x
```

If you declare a variable inside the loop, then it is only available inside the loop.

```julia
xoutsideloop=3.0

for i=1:3
    xoutsideloop+=1.0
    xinsideloop=xoutsideloop+1.0
end
```

```julia
xoutsideloop
```

Once you finish executing the loop, the program outside the loop does not know anything about `xinsideloop` which was declared inside the for loop.

```julia
xinsideloop
```
